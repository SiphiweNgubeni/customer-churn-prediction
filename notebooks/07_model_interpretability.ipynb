# 07 - Model Interpretability (Phase 9)

Use SHAP to explain the selected model. Feature importance indicates association, not causation.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import shap
from src.predict import load_model
from src.data_preprocessing import load_raw_data, clean_data, split_features_target
from src import config

pipe = load_model()
X, y = split_features_target(clean_data(load_raw_data()))
pre = pipe.named_steps['preprocess']
model = pipe.named_steps['model']
sample = X.sample(min(len(X), 400), random_state=config.RANDOM_STATE)
X_trans = pre.transform(sample)
names = list(pre.get_feature_names_out())
```

```python
explainer = shap.Explainer(model, X_trans, feature_names=names)
shap_values = explainer(X_trans)
shap.summary_plot(shap_values, features=X_trans, feature_names=names)
```

```python
# Explain a single customer
shap.plots.waterfall(shap_values[0])
```
