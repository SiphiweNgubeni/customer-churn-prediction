## Chi-square: is Contract associated with Churn?
H0: Contract and Churn are independent. H1: they are associated.

```python
res = chi_square_test(df, 'Contract')
print(f"chi2={res['chi2']:.2f}  p={res['p_value']:.3e}  Cramer's V={res['cramers_v']:.3f}")
print('Reject H0 at 5%' if res['significant_at_0.05'] else 'Fail to reject H0')
```

```python
for f in ['PaymentMethod','InternetService','TechSupport']:
    r = chi_square_test(df, f)
    print(f"{f:>16}  p={r['p_value']:.3e}  V={r['cramers_v']:.3f}")
```
