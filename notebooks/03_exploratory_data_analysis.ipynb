# 03 - Exploratory Data Analysis (Phase 4)

Discover patterns associated with churn. Compare churn **rates** within groups, not just raw counts.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
from src.data_preprocessing import load_raw_data, clean_data
df = clean_data(load_raw_data())
```

```python
df['Churn'].value_counts(normalize=True)
```

```python
sns.countplot(data=df, x='Churn')
plt.title('Customer Churn Distribution'); plt.show()
```

```python
sns.countplot(data=df, x='Contract', hue='Churn')
plt.title('Churn by Contract Type'); plt.xticks(rotation=20); plt.show()
```

```python
# Churn RATE by contract (not raw counts)
tmp = df.assign(flag=df['Churn'].map({'No':0,'Yes':1}))
tmp.groupby('Contract')['flag'].mean().sort_values(ascending=False)
```

```python
sns.boxplot(data=df, x='Churn', y='tenure'); plt.title('Tenure by Churn'); plt.show()
sns.boxplot(data=df, x='Churn', y='MonthlyCharges'); plt.title('Monthly Charges by Churn'); plt.show()
```
